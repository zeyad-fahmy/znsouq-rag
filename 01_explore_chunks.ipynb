```python
from rag.loader import load_chunks

chunks = load_chunks()
print("Total chunks:", len(chunks))
```

```python
lengths = [len(c["text"]) for c in chunks]
print("Average length:", sum(lengths) // len(lengths), "characters")
print("Shortest:", min(lengths), "| Longest:", max(lengths))
```

```python
for c in chunks:
    if c["metadata"]["source"] == "platform/buyer/returns-policy.md":
        print("=" * 80)
        print(c["metadata"]["section"], "|", c["metadata"]["source"])
        print("-" * 80)
        print(c["text"][:600])
```

```python
store = [c for c in chunks if c["metadata"].get("seller_id") == "S0017"]
for c in store:
    print(c["metadata"])
    print(c["text"][:300], "\n")
```
