```python
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer("BAAI/bge-small-en-v1.5")
a, b, c = model.encode([
    "Can I send back a shirt I already washed?",
    "Items must be unworn, unwashed and with original tags attached",
    "Orders to Cairo arrive in 1-2 business days",
])

def cosine(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

print("question vs return rule:", round(float(cosine(a, b)), 3))
print("question vs shipping:   ", round(float(cosine(a, c)), 3))
```
