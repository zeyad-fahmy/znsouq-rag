```python
import os
os.environ["HF_HUB_OFFLINE"] = "1"
from rag.answer import answer

def ask(q, mode="buyer"):
    r = answer(q, mode)
    print("Q:", q, f"  [{mode}]", f"(store: {r['seller_id']})" if r["seller_id"] else "")
    print("-" * 80)
    print(r["answer"])
    print("\nSources:")
    for i, h in enumerate(r["sources"], 1):
        print(f"  [{i}] {h['score']}  {h['metadata']['source']} › {h['metadata']['section']}")
    print("=" * 80, "\n")
```

```python
ask("Can I return a shirt I already washed?")                                   # basic fact
ask("I bought headphones from Falcon Deals 10 days ago. Can I still return them?")  # store vs. platform conflict
ask("When do I get paid?", mode="seller")                                       # audience
ask("What is the referral fee for Electronics?", mode="seller")                 # date-dependent
ask("Can I pay with Bitcoin?")                                                  # not supported
ask("What is the capital of France?")                                           # out of scope
```
